## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

I also checked that the two 45-day windows use past data only. The first window is the previous 45 days and the second window is the most recent 45 days. No future data is used in the score. The project does not contain product flags or other future information.
