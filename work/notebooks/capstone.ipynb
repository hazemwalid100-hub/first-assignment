## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

In my current project, I built a rule-based scoring engine rather than a trained machine learning model. Therefore, I cannot honestly claim that a machine learning model performed better than a baseline yet. The current results show that the scoring engine processed 200 page and keyword combinations and produced 131 actionable recommendations.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Results from the project pipeline run
results = pd.DataFrame({
    "Action": [
        "CTR optimization",
        "Decay refresh",
        "Monitor stable"
    ],
    "Count": [99, 32, 69]
})

results["Percentage"] = (
    results["Count"] / results["Count"].sum() * 100
).round(1)

print("Project Results")
print("---------------")
print("Total combinations processed: 200")
print("Actionable recommendations: 131")
print("Highest composite score: 45.73")
print("Highest-ranked action: ACT_DECAY_REFRESH")
print("Highest traffic decline: 51.3%")
print("Position deterioration: 4.1 spots")
print()

display(results)

plt.figure(figsize=(8, 4))
bars = plt.bar(results["Action"], results["Count"],
               color=["#b6492d", "#d97b55", "#b9b6ae"])

plt.title("Project Action Recommendations")
plt.ylabel("Number of page/keyword combinations")
plt.xticks(rotation=15)
plt.grid(axis="y", alpha=0.25)

for bar, value in zip(bars, results["Count"]):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 2,
        str(value),
        ha="center"
    )

plt.tight_layout()
plt.show()
```
